## Framework penulisan Python
Sebenarnya tidak ada kerangka khusus (framework) dalam pemrograman python. Hanya saja, programmer Python umumnya menggunakan kerangka atau bentuk penulisan kode program sebagai berikut: 

    def main():
        # daftar perintah
        . . .
        
    if __name__ == "__main__":
        main()
        
Bentuk di atas, pertama kali adalah mendefinisikan fungsi main() dan kemudian fungsi tersebut akan dipanggil dari modul utama (__main__). Perhatikan contoh di bawah ini

```python
# -*- coding: utf-8 -*-
"""
 @author Priyandari
 @create date 2020-05-04 23:57:23
 @modify date 2020-05-04 23:57:23
 @desc [description]
"""

import os

# Membuat fungsi main()
def main():
    program = ['Python', 'Go', 'Rubby', 'Perl', 'PHP']
    for p in program:
        print(p)
    os.system("pause");
    
if __name__ == "__main__":
    main()
```
