## 1. Membaca File .csv dan .xlsx

```python
import pandas as pd

filepath = 'files/TI191107_2022_Ganjil_A.xlsx'

df_xls = pd.read_excel(
    filepath,
    skiprows=14,
    header=None,   # tidak menggunakan header dari file
    names=["Nim", "Mahasiswa", "UK3", "UK4", "UTS", "UAS", "UK5"],
    usecols=[1,2,3,4,5,6,7],
    sheet_name="Sheet1"
)

print(df_xls.head(2))
# df_xls.head(2)
```

## 2. Menulis File .csv dan .xlsx



```python
# Menyimpan file ke CSV
file_path_csv = "files/Nilai_KelasA.csv"
df_xls.to_csv(file_path_csv, sep=";",index=False)
print(f"Berhasil di simpan ke {file_path_csv}")

# Menyimpan file ke XLSX
file_path_xls = "files/Nilai_KelasA.xlsx"
df_xls.to_excel(file_path_xls, index=False)
print(f"Berhasil di simpan ke {file_path_xls}")

# Menyimpan file ke XLSX dengan nama sheet tertentu
with pd.ExcelWriter(file_path_xls, engine="openpyxl") as writer:
    # df_xls.to_excel(writer, sheet_name="Nilai_A", index=False)
    df_xls.to_excel(writer, sheet_name="Nilai_A", index=False)
    df_csv.to_excel(writer, sheet_name="Nilai_C", index=False)
    print(f"Berhasil di simpan ke {file_path_xls} dengan sheet tertentu")
```
