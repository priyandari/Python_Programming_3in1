## 5. PIP

pip adalah paket library, yang dipkai untuk melakukan proses instalasi library-libarry yang dibutuhkan program python. 

```python
# Check versi pip pada python default dan python versi tertentu (misal ada lebih dari 1 python dalam komputer)
pip --version
py -3.10 -m pip --version

# Daftar Packet
pip list
py -3.10 -m pip list

# Install library
pip install dowhy

# Install library versi tertentu
pip install pandas==2.3.0
 
# Upgrade library ke versi tertentu
pip install --upgrade dowhy==0.13
```
