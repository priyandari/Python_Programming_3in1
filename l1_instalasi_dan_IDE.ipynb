## 1. Instalasi

Silahkan unduh program Python pada alamat url http://python.org
Tutorial kali ini menggunakan Python 3 dengan versi 3.9.x atau yang lebih baru.

Selain menggunakan file instalasi dari http://python.org, Anda bisa juga melakukan instalasi menggunakan Anaconda https://www.anaconda.com/download/. Anaconda menyediakan program python dan juga sejumlah library yang cukup lengkap untuk data analytics.  

#### untuk OS  mac
https://www.python.org/downloads/mac-osx/

Disarankan menggunakan python versi 3.9.x ke atas, atau langsung the latest juga boleh.

#### Install dua VERSI berbeda dalam 1 PC

Buka python.org/downloads

Pilih versi Python yang kamu butuhkan (3.10, atau lainnya).

Download installer .exe dan install.

⚠️ Saat instalasi, jangan centang "Add Python to PATH" untuk semua versi sekaligus, agar tidak bentrok.

Akses dengan:

`py -3.10`

`py -3.11`
