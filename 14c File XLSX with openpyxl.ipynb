## 1. Membaca File .xlsx

```python
import openpyxl as pyxl
# Path, Nama file, dan sheet
file_path = "files/TI191107_2022_Ganjil_A.xlsx"
wb = pyxl.load_workbook(file_path, data_only=True)
sheet = wb["Sheet1"]
```

```python
# Membaca dan menampilkan hasil
for row in sheet.iter_rows(min_row=15, max_row=20, min_col=2, max_col=7):
    print([cell.value for cell in row])
```

2. Update beberapa nilai mahasiswa sekaligus dimana data berupa dictionary dengan key NIM, dan nilainya juga dictionary dengan indeks kolom dan nilai baru


nilai_baru_mhs = {

```python
# Update beberapa mahasiswa dan beberapa  nilai dan kolom tertentu

nilai_updates = {
    17: {4:65,5:60,6:60,7:65},
    19: {5:65,7:90},
    20: {5:60}
}

for baris, koloms in nilai_updates.items():
    print("nomor ", baris, " kolom ", koloms)
    for col, val in koloms.items(): 
        sheet.cell(row=baris, column=col).value = val

# Simpan perubahan
wb.save(file_path)
print("Update berhasil")
```

```python
# Lihat hasil update
print("Hasil update")
wb = pyxl.load_workbook(file_path, data_only=True)
sheet = wb["Sheet1"]
# Membaca dan menampilkan hasil
for row in sheet.iter_rows(min_row=15, max_row=20, min_col=2, max_col=7):
    print([cell.value for cell in row])
```

```python
# NIM dicari
nilai_baru_mhs = {
    "I0322002": {6:90},
    "I0322004": {6:96,7:96,},
    "I0322006": {6:85}
}

# Kolom NIM --> B = 2)
kolom_target = 2

indeks_baris = None

# Loop seluruh baris di kolom tersebut
for key, koloms in nilai_baru_mhs.items():
    for row in sheet.iter_rows(min_row=15, max_col=20, min_col=kolom_target):
        cell = row[0]
        if cell.value == key:
            indeks_baris = cell.row
            # print(indeks_baris)
            for col, val in koloms.items(): 
                sheet.cell(row=indeks_baris, column=col).value = val
            break

# Simpan perubahan
wb.save(file_path)
print("Update berhasil")
```
